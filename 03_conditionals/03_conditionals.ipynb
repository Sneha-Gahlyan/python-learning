#Conditional Statements

```python
# Basic if statement
temperature = 35
if temperature > 30 :
  print("It is a hot day.")


#if-else statement
number = 10
if number % 2 == 0 :
  print("The number is even.")
else :
  print("The number is odd.")

#if-elif-else
marks = 82

if marks >= 90 :
  print("Grade A")
elif marks >= 80 :
  print("Grade B")
elif marks >= 70 :
  print("Grade C")
else :
  print("Grade D")
```

# Practice Example

```python
# Example 1 : Output ?
x = 10
if x > 5 :
  print("x is greater than 5")
elif x > 8 :
  print("x is greater than 8")
else:
  print("x is not greater than 5 and 8")

#Both x > 5 and x > 8 are technically true.
#But Python checks: x > 5 → True
#So it executes A and doesn't check the elif.

#Example 2 : Nested condition
age = 20
has_id = True
if age > 18 :
  if has_id :
    print("Entry Allowed")
  else:
    print("Entry Denied")
else:
  print("underage")

#Example 3 : and condition
age = 20
has_id = True
if age > 18 and has_id :
  print("Entry Allowed")
else:
  print("Entry Denied")

# Example 4 : Truthy/Falsy
name =""
if name :
  print("Name exists")
else:
  print("Name is empty")

#Example 5 : 0 falsy
x = 0

if x:
    print("True")
else:
    print("False")
```
