#Attribute

```python
class Student1:
  name = "Sneha" #class attribute

  def __init__(self,age):
    self.age = age #instance attribute

object1 = Student1(22)
print(object1)
print(f"Student name and age is {object1.name} and {object1.age}")

# object1.name → accesses name
# object1.age → accesses age
# object1 → refers to the whole object
```

#Methods

```python
# instance methods: it works with instance of the class
# class method : it works with the class itself


class Student2:

  def __init__(self , name , age):
    self.name = name
    self.age = age

  def show(self) :  #instance method
    print(f"Student name and age is : {self.name} and {self.age}")

  @classmethod
  def college(cls,clgname):
    cls.clg = clgname
    print(f"College name of the student is : {cls.clg}")

  @staticmethod
  def is_adult(ages):
    return ages>=18

object2 = Student2("Sneha",22)
object2.show()

Student2.college("YMCA Faridabad")  #classmethod

print(Student2.is_adult(13)) #static method
print(Student2.is_adult(22))
```
