# Encapsulation

Encapsulation is the practice of bundling attributes and the methods that operate on that data inside a class, while controlling how that data can be accessed or modified.

# Why do we need Encapsulation?

```python
class BankAccount:

  def __init__(self,balance):
    self.balance = balance


account = BankAccount(10000)

account.balance = -500

print(account.balance)
```

# Access Modifiers


```python
# Public Attributes and Methods

# A public attribute can be accessed directly from outside the class.

class Student:

    def __init__(self, name):
        self.name = name  # Public attribute


student = Student("Sneha")

print(student.name)
```

```python
# Protected Members and Methods

""" python protected members are created using a single
underscore but it still can be accessed from outside the
class
A single underscore represents a convention, not strict access restriction."""

class Student:

    def __init__(self, name):
        self._name = name


student = Student("Sneha")

print(student._name)


```

```python
# Private Members

# Python uses a double underscore __ to indicate a private member.

class BankAccount:

    def __init__(self, balance):
        self.__balance = balance


account = BankAccount(10000)

print(account.__balance)
```

# Accessing Private Data Through Methods

```python
# Instead of accessing __balance directly, we can provide methods to control access.

class BankAccount:

  def __init__(self,balance):
    self.__balance = balance

  def get_balance(self):
    return self.__balance


account = BankAccount(5000)

print(account.get_balance())
```
