#Polymorphism

It allows the same interface or method name to behave differently
depending on the object or context.

```python
# Method Overriding

class Animal :

  def sound(self):
    print("Animal makes a sound")


class Dog(Animal):

  def sound(self):
    print("Dog barks")


class Cat(Animal):

  def sound(self):
    print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

# Duck Typing

“If it walks like a duck and quacks like a duck, it must be a
duck"

If an object has the required method, Python can use that method—regardless of the object's class.

```python
class Dog:

    def sound(self):
        print("Dog barks")


class Cat:

    def sound(self):
        print("Cat meows")


def make_sound(animal):
    animal.sound()


dog = Dog()
cat = Cat()

make_sound(dog)
make_sound(cat)
```
