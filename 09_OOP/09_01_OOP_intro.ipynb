#OOP

OOP is a programming paradigm based on objects

Objects combine:
    - Data → Attributes
    &
    - code → Methods

```python
class Addition:
    """A simple class demonstrating the basic idea of OOP."""

    def __init__(self, a, b):
        self.a = a
        self.b = b

    def calculate(self):
        return self.a + self.b


# Creating an object
addition = Addition(12, 12)

# Calling a method
result = addition.calculate()

print("Result:", result)
```

#Classes and Objects

Classes are the blueprint for creating an object  
and object is instance of an class

```python

# Employee : Class
# employee1: object
# employee2: object

class Employee:
    """Represents an employee."""

    def display_info(self):
        print("Employee information")


# Creating objects from the Employee class
employee1 = Employee()
employee2 = Employee()

# Calling a method using objects
employee1.display_info()
employee2.display_info()
```

#Attributes

```python
"""
Attributes are variables that store data associated
with an object.

Instance attributes belong to individual objects.
"""


class Employee:

    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.salary = salary


# Creating objects with different attribute values
employee1 = Employee("Rahul", "Sales", 45000)
employee2 = Employee("Priya", "HR", 50000)

# Accessing attributes
print(employee1.name)
print(employee1.department)
print(employee1.salary)

print(employee2.name)
print(employee2.department)
print(employee2.salary)
```

#Methods

```python
"""
A method is a function defined inside a class.
Methods define the behavior of objects.

Attribute → name, salary

Method → display_info()
Method → annual_salary()
"""


class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_info(self):
        print(f"Employee: {self.name}")
        print(f"Salary: {self.salary}")

    def annual_salary(self):
        return self.salary * 12


# Creating an object
employee = Employee("Rahul", 50000)

# Calling methods

employee.display_info()

yearly_salary = employee.annual_salary()

print("Annual Salary:", yearly_salary)
```

#Constructor

```python
"""
The __init__() method is automatically called when
a new object is created.

It is commonly used to initialize object attributes.
"""


class Product:

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_value(self):
        return self.price * self.quantity


# Creating an object
product = Product("Laptop", 55000, 2)

print("Product:", product.name)
print("Price:", product.price)
print("Quantity:", product.quantity)
print("Total Value:", product.total_value())
```

#Bag factory example:
in this we want material,zips and pockets from user to create object

```python
class Factory:
  def __init__(self,material,zips,pockets) -> None:
      self.material = material
      self.zips = zips
      self.pockets = pockets

  def show(self) :
    print(f"The Object details are : Material = {self.material}\nPockets = {self.pockets} \nzips = {self.zips}")

reebok = Factory("Leather",3,2)

reebok.show()
```
