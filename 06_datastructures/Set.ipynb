#Set

```python
# Set is an unordered collection of unique elements

numbers = {1,2,3,4,5,6}
print(numbers)

# No duplicates
a = {1,3,3,4,4,4,4,4}
print(a)

# No indexing
# a[0]  is invalid


#Each value in a set is hashed using a hash function , hash()
# The hash is used as an index to store the element in memoryOy

b = hash("Hello")
print(b)
c = hash((1,2,3,44))
print(c)
```

#Traversing


```python
# Traverse in set using loop
a = {1,4,8,9,10,11,12,2}
for i in a :
  print(i)
```

# Set methods


```python
n = {1,3,4,6,7}

# 1. Add elements
n.add(10)
print(f"Add element 10 : {n}")

# 2. remove()

# If the element doesn't exist, remove() raises a KeyError.

n.remove(4)
print(f"Remove 4 : {n}")

# 3.discard()

# If 5 doesn't exist, no error occurs.
n.discard(5)
print(f"After discarding 5 : {n}")


# 4. pop() : removes and returns an arbitrary element from the set.

# Don't rely on a particular element being removed.

removed = n.pop()
print(f"Removed {removed} from set and set is {n}")


# 5. clear() : it removes all elements
n.clear()
print(n)
```

# Set Operations

```python
a = {1,2,3,4,5}
b = {4,5,6,7,8}

# Union combines all unique elements from both sets.
union_set = a.union(b)
print(union_set)
print(a|b) #union

# Intersection returns the elements that are common to both sets.
intersection_set = a.intersection(b)
print(intersection_set)
print(a&b) # intersection

# Elements present in A but NOT present in B.
difference_set = a.difference(b)
print(difference_set)
print(a-b) #difference set : elements only in A

# Symmetric Difference : Returns elements that are in either set, but not in both.
symmetric_diff = a.symmetric_difference(b)
print(symmetric_diff)
print(a^b) # symmetric difference

```
