#Tuple

```python
a= (1,2,3,5,6)

print(type(a))
```

#Tuple traversing

```python
# Traverse using index
a = (1,5662,1,56,6)

for i in range(len(a)):
  print(a[i])

print()

# Traverse directly on values
for i in a:
  print(i)
```

# Tuple Methods


```python
b = (5,24,5,557,7,54,5,3,557)
# 1. index() : find the index of first occurence
index = b.index(5)
print(index)

# 2. count() : count occurence of element
count = b.count(5)
print(count)
```

#Tuple defining

```python
a,b,c,d = (1,2,3,4)
print(c)

m = (90)
print(type(m))

# for single value tuple
n = (24,)
print(type(n))
```
