#Traversing

```python
a = {10:100,20:200,30:300}

# access keys
for i in a :
  print(i)

print()

# access values
for i in a :
  print(a[i])

print()

# access directly values
for i in a.values():
  print(i)
```

#Dictionary Questions

```python
# 1. Write a Python script to merge two Python dictionaries

a = {10:100,20:200,30:300}
b = {40:400,50:500,60:600}

for i in b:
  a[i] = b[i]

print(a)


```

```python
# 2.  Write a Python program to sum all the values in a dictionary

d = {10:100,20:200,30:300}
sum = 0
for i in d:
  sum = sum + d[i]

print(sum)
```

```python
# 3.  Count the frequency of each element

n = [1,3,5,7,8,9,1,1,1,2,5,1,2,3,4,8,9,9,8,2,3,7]

d = {}

for i in n :
  if i in d.keys() :
    d[i] += 1
  else:
    d[i] = 1

print(d)
```

```python
# 4. Write a Python program to combine two dictionary by adding
# values for common keys.


d1 = {10:100,20:200,30:300 , 40:400}
d2 = {40:200,50:500,60:600}

for i in d2:
  if i in d1.keys():
    d1[i] += d2[i]
  else:
    d1[i] = d2[i]

print(d1)
```
