#List Methods

```python
numbers = [3,56,8,2,6,2,23]

# 1. append() : Add one element at the end of the list
numbers.append(98)
print(f"append(98) : {numbers}")

# 2. insert() : Add an element at a specific index
numbers.insert(2,76)
print(f"index(2,76) : {numbers}")

# if the index is greateer than the list length , python puts the element at the end
a = [10,3,65]
a.insert(100,40)
print(f"Trick q : {a}")

# 3. extend() : add multiple elements to the end of list
numbers.extend([45,22,40])
print(f"extend([45,22,40]) : {numbers}")

# 4. remove() : remove the first occurence of a specified value
numbers.remove(2)
print(f"remove(2) : {numbers}")
#if value doesn't  exist than give ValueError

# 5. pop() : Removes an element using its index and returns that removed element.
poped = numbers.pop(6)
print(f"The remove element is {poped} and list is {numbers}")

# 6. index() : Finds the index of the first occurrence of a value.
position = numbers.index(6)
print(f"The position of 6 is {position}")

# 7. count() : Counts how many times a value occurs in a list.
count = numbers.count(40)
print(f"The count of 40 is {count}")

# 8. sort() : Sorts the list in ascending order by default.

#  sort() modifies the list in-place and does not return the sorted list.

numbers.sort()
print(f"Ascending order : {numbers}")

numbers.sort(reverse=True)
print(f"Descending order : {numbers}")


# 9. reverse() : Reverses the current order of elements.
numbers.reverse()
print(f"Reversed list : {numbers}")

# 10. copy() : Creates a copy of the list
copy_numbers = numbers.copy()
print(f"Copied list : {copy_numbers}")


# new_numbers = numbers
#new_numbers = numbers.copy()
# = creates another reference to the same list, while copy() creates a separate list (a shallow copy).

#11. clear() : Removes all elements from the list.
numbers.clear()

print(numbers)
```

#List Practice Questions

```python
# 1. Print positive and negative elements of an List?
numbers = [10, -5, 7, -2, 0, 15, -8]

positive = []
negative = []

for i in numbers:
  if i>=0:
    positive.append(i)
  elif i<0:
    negative.append(i)

print(f"Positive elements : {positive}\nNegative Elements : {negative}")

```

```python
# 2.  Mean of List elements
num = [2,4,6,74,341,1]

sum =0

for i in num:
  sum += i

Mean = sum/len(num)

print(Mean)
```

```python
# 3. Find the greatest element and print its index too?

l = [12,574,3,4,56,332]

largest = l[0]
index = 0

for i in range(0,len(l),1):
  if largest < l[i] :
    largest = l[i]
    index = 0
print(f"Largest is {largest} and index is {index}")
```

```python
# 4. Find the second greatest element?

l = [12,574,3,4,56,332]

largest = l[0]
sec_largest = l[0]


for i in l:
  if i>largest:
    sec_largest = largest
    largest = i
  elif i> sec_largest:
    sec_largest = i


print(f"Largest: {largest} \nSecond Largest : {sec_largest}")
```

```python
# 5.  Check if List is sorted or not

sorting = [2,5,8,10,15]

for i in range(len(sorting)-1):
  if sorting[i] < sorting[i+1]:
    continue
  else:
    print("List is not sorted")
    break
else:
  print("List is sorted")
```
