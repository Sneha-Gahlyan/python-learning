#Functions
A function is a reusable block of code that performs a specific task .

```python
def greet():
  print("Hello,Python!")

greet()
```

#Types of functions

```python
#Built-in function
#These are functions already provided by Python

name = "Sneha"
print(name)
print(len(name))

#User-defined functions
#functions created by the programmer using def.
def greet():
  print("Hello,Python!")

greet()
```

#Parameters and Arguments

```python
#Parameters: it us a variable weritten inside the function definition
def greet(name):
  print(f"Hello , {name}")

greet("sneha")
# Here name is the parameter

#Argument : The actual value passed when calling the function is an argument

# Here sneha is the argument
```

#Types of arguments

```python
#1. Positional argument
# Arguments are matched according to their position
def student (name,age):
  print(f"Name: {name} \n Age: {age}")

student("Sneha",22)
```

```python
#2. Default argument
# A parameter can have a default value
def greet(name = "Guest"):
  print(f"hello, {name}")

greet("Sneha") #replace default
greet() #default use
```

```python
#3. Keyword Arguments
def student(name,age):
  print(f"Name: {name} \n Age: {age}")

student(age = 22, name = "Sneha")
```

#Functions with a return value

```python
# return send a value back to the place where the function was called
def add(a,b):
  return a+b

result = add(10,20)

print(result)
```
