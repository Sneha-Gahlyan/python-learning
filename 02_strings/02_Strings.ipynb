#02_Strings


#String

```python
# Creating strings

name = "Sneha"
course = 'Python'
message = """Learning Python step by step."""

print(name)
print(course)
print(message)
print(type(name))
print(type(course))
print(type(message))
```

#Unicode

```python
character = "A"

print(ord(character))

print(chr(65))
```

#Indexing


```python
#  S   n   e   h   a
#  0   1   2   3   4
# -5  -4  -3  -2  -1


name = "Sneha"

# Positive indexing
print(name[2])

# Negative indexing
print(name[-2])
```

#String Slicing

```python
name = "Sneha Gahlyan"

print(name[6:13:1])


# Reverse a string

print(name[::-1])
```
