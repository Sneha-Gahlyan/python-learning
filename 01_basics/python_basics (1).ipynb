#Data Types


```python
#Numbers

#Integer
year = 2026
print(year,type(year))

#Float
pi = 22/7
print(pi,type(pi))

#Complex
complex = 2+3j
print(complex,type(complex))

#String
name = "Sneha"
course = 'Btech computer engg.'

print(name,type(name))
print(course,type(course))

#Boolean
is_student = True
is_teacher = False

print(is_student,type(is_student))
print(is_teacher,type(is_teacher))

```

#Type Converrsion

```python
#int() : converts a value into int
year = '2026'
year = int(year)
print(year,type(year))

# Float -> integer
pi = 22/7
pi = int(pi)
print(pi,type(pi))

#float() : converts a value into float
age = 22
age = float(age)
print(age,type(age))

#str() : converts a value into string
age = 22
age = str(age)
print(age,type(age))

#bool() : converts a value into boolean
student_rollno = 24
student_rollno = bool(student_rollno)
print(student_rollno,type(student_rollno))

student_marks = 0
student_marks = bool(student_marks)
print(student_marks,type(student_marks))
```
