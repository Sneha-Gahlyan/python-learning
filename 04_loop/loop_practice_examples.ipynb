```python
# 7. Print the sum of all even & odd numbers in a range separately
e = int(input("Enter an integer: "))
even = 0
odd = 0
for i in range(1,e+1,1):
  if i%2==0:
    even = even+i
  else:
    odd = odd+i
print(f"The even sum is {even} and odd sum is {odd}")
```

```python
# 9. - Accept a number and check if it a perfect number or not.
# A number whose sum of factors is equal to the number itself
# Ex - 6 = 1, 2, 3 =
g = int(input("Enter a number : "))
sum = 0
for i in range(1,g,1):
  if g%i == 0 :
    sum += i

if sum == g:
  print(f"{g} is a perfect number")
else:
  print(f"{g} is not a perfect number")
```

```python
# 10. Check wether the number is prime or not
f = int(input("Enter a number : "))
count =0
for i in range(1,f+1,1):
  if f%i == 0:
    count += 1

if count  == 2:
  print(f"{f} is a prime number")
else:
  print(f"{f} is not a prime number")
```

#Loop on string


```python
# 11. Print a Reverse  string without using in build functions.

g = "Sneha Gahlyan"

for i in range(len(g)-1 , -1, -1):
  print(g[i])
```

```python
# 12. Reverse a string without using in build functions.

h = "Sneha Gahlyan"
rev = ""
for i in range(len(h)-1 , -1, -1):
  rev += h[i]

print(rev)
```

```python
#13. Check string is Pallindrome or not

j = input("Enter a string to check palindrome : ")
rev = ""

for i in range(len(j)-1 , -1, -1):
  rev += j[i]

if rev == j:
  print("It is a pallindrome")
else:
  print("It is not a pallindrome")
```

```python
# 14. Count all letters, digits, and special symbols from a given string
# Given: str1 = "P@#yn26at^&i5ve"
# Expected Outcome: Total counts of chars, digits, and symbols
# Chars = 8
# Digits = 3
# Symbol = 4

k = input("Enter a string :")
chars  =0
digits =0
symbol =0
for i in k:
  if i.isdigit():
    digits += 1
  elif i.isalpha():
    chars += 1
  else:
    symbol += 1

print(f"Chars = {chars} \n Digits = {digits} \n Symbol = {symbol}")

```

#While loop


```python
# 15. Separate each digit of a number and print it on the new line
l = int(input("Enter a number : "))

while l>0:
  print(l%10)
  l //= 10
```

```python
#16. Accept a number and print its reverse
m = int(input("Enter a number : "))

rev =0
while m>0 :
  rev = rev * 10 + m%10
  m = m//10
print(rev)
```

```python
# 17. Accept a number and check if it is a pallindromic number (If
# number and its reverse are equal?

n = int(input("Enter a integer:"))
copy  = n
reverse = 0
while n>0:
  reverse = reverse * 10 + n%10
  n //= 10

if reverse == copy:
  print(f"{copy} is a pallindrome")
else:
  print(f"{copy} is not a pallindrome")
```

#Game using while loop

```python
# 18. Create a random number guessing game with python.
import random
num = random.randint(1,10)

tries = 0

while True :
  guess = int(input("Guess a number between 1 to 10 : "))
  tries += 1
  if num == guess :
    print(f"You won, you guessed the number in {tries} tries")
    break
  elif num > guess :
    print("guess little higher number")
  else:
    print("guess little lower number")
```
