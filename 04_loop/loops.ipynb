#Loop

# For Loop for numbers

```python
# print 1-20 numbers with for loop
a = range(1,21,1)
for i in a :
  print(i)

# By giving only stop condition in range
b = range(21)
for i in b :
  print(i)
```

```python
# print 16-1
a = range(16,0,-1)
for i in a :
  print(i)
```

```python
#print a table of 5
a = range(5 , 51 , 5)
for i in a :
  print(i)
```

```python
# print table acc to entered number
num = int(input("Enter a number for table :"))
x = range(num , (num *10)+1 , num )
for i in x:
  print(i)

# print table in different way
for i in range(1,11):
  print(f"{num} * {i} = {num*i}")
```

#Loop for Strings

```python
#using index values
a = "Sneha"
for i in range(0,len(a),1):
  print(a[i])

for i in range(len(a)):
  print(a[i])
```

#Iterating directly over the string

```python
name = "Sneha"
for i in name:
  print(i)
```

#Break


```python
# break : exit a loop permanently if a certain condition is met
#if a break statement encountered ,loop stops immendiately and control move to next statement after loop

#print 1-15 : if after 10 there should not be printed than break
for i in range(1,16,1):
  if(i==90):
    print("Break statement is executed")
    break
  else:
    print(i)
else :
  print("Break statement is not executed")
```

#Continue

```python
#continue: skips one of the statement used with if and run rest of the iteration will run

for i in range(1,16,1):
  if(i==10):
    print("Continue statement is executed")
    continue
  else:
    print(i)
```
